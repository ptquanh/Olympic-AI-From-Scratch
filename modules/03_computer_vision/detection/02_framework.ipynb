# NMS bằng torchvision
Sử dụng API chuẩn để tối ưu tốc độ.


```
import torch
import torchvision.ops as ops

boxes = torch.tensor([[10, 10, 50, 50], [12, 12, 52, 52], [100, 100, 150, 150]], dtype=torch.float)
scores = torch.tensor([0.9, 0.8, 0.7])

keep = ops.nms(boxes, scores, iou_threshold=0.5)
print('Boxes được giữ lại:', keep.tolist())
```
