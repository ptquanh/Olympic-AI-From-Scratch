```
from torchvision.ops import box_iou, nms

boxes = torch.tensor(
    [[0.0, 0.0, 10.0, 10.0], [1.0, 1.0, 9.0, 9.0], [20.0, 20.0, 30.0, 30.0]]
)
scores = torch.tensor([0.9, 0.8, 0.7])
matrix = box_iou(boxes, boxes)
keep = nms(boxes, scores, 0.5)
assert keep.tolist() == [0, 2]
print(matrix, keep)
```
