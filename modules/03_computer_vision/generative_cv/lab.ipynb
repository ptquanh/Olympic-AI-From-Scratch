# Generative CV — forward diffusion toy

Demo noise schedule không cần model hoặc package ngoài profile.

```python
x0=np.linspace(-1,1,64).reshape(8,8); rng=np.random.default_rng(42); noise=rng.normal(size=x0.shape)
for alpha_bar in (1.,.5,.05):
    xt=math.sqrt(alpha_bar)*x0+math.sqrt(1-alpha_bar)*noise
    print(alpha_bar,round(float(xt.mean()),3),round(float(xt.std()),3))
assert np.allclose(math.sqrt(1.)*x0+math.sqrt(0.)*noise,x0)
```

**Observation:** khi `alpha_bar` giảm, tín hiệu gốc yếu dần. Reverse diffusion học dự đoán noise/score để đảo quá trình này.
