# Tự code ConvTranspose2d (Phóng to ảnh)

```
import torch
import torch.nn as nn

# Tensor 1x1
x = torch.randn(1, 1, 2, 2)
up = nn.ConvTranspose2d(1, 1, kernel_size=2, stride=2)
out = up(x)
print('Original:', x.shape)
print('Upsampled:', out.shape) # Sẽ là 4x4
```
