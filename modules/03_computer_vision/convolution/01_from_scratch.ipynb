# Conv2D NumPy

Deep-learning libraries implement cross-correlation (không lật kernel) nhưng thường gọi là convolution.

```
def conv2d(x, kernel, stride=1, padding=0):
    """Cross-correlation for x[H,W], kernel[kH,kW] -> out[oH,oW]."""
    x = np.pad(np.asarray(x, float), padding)
    k = np.asarray(kernel, float)
    oh = (x.shape[0] - k.shape[0]) // stride + 1
    ow = (x.shape[1] - k.shape[1]) // stride + 1
    if oh <= 0 or ow <= 0:
        raise ValueError("kernel is larger than padded input")
    out = np.empty((oh, ow))
    for i in range(oh):
        for j in range(ow):
            patch = x[
                i * stride : i * stride + k.shape[0],
                j * stride : j * stride + k.shape[1],
            ]
            out[i, j] = np.sum(patch * k)
    return out


x = np.arange(1, 17).reshape(4, 4)
k = np.array([[1, 0], [0, -1]])
out = conv2d(x, k, stride=2)
assert out.shape == (2, 2)
print(out)
```
