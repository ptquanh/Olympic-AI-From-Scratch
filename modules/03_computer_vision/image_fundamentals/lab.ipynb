# Load Image with OpenCV

```
import cv2
from matplotlib import pyplot as plt

# Tạo ảnh dummy (đen)
img = np.zeros((100, 100, 3), dtype=np.uint8)
img[:, 50:] = [255, 0, 0] # Nửa phải màu BGR (Xanh dương)

# Vẽ ảnh (Hiển thị màu sai nếu không convert)
plt.imshow(img)
plt.title('BGR Image (Matplotlib thinks it is RGB)')
plt.show()
```
