# Thử nghiệm Torchvision Transforms

```
import torchvision.transforms as T
import torch

# Dummy Tensor mô phỏng ảnh
img_tensor = torch.randn(3, 100, 100)

# Transform
t = T.RandomHorizontalFlip(p=1.0)
out = t(img_tensor)
print('Output shape:', out.shape)
```
