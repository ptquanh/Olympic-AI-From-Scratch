# Experiment — vì sao chia sqrt(dk)?

```
def entropy(p):
    return -(p * np.log(p + 1e-12)).sum(-1).mean()


rng = np.random.default_rng(42)
for d in (8, 64, 512):
    q = rng.normal(size=(64, d))
    k = rng.normal(size=(64, d))
    raw = q @ k.T
    print(
        d,
        "unscaled",
        round(
            entropy(
                np.exp(raw - raw.max(1, keepdims=True))
                / np.exp(raw - raw.max(1, keepdims=True)).sum(1, keepdims=True)
            ),
            3,
        ),
        "scaled",
        round(
            entropy(
                np.exp(raw / math.sqrt(d) - (raw / math.sqrt(d)).max(1, keepdims=True))
                / np.exp(
                    raw / math.sqrt(d) - (raw / math.sqrt(d)).max(1, keepdims=True)
                ).sum(1, keepdims=True)
            ),
            3,
        ),
    )
```

**Observation:** không scale làm logits tăng variance theo `d_k`, softmax dễ bão hòa và entropy giảm.
