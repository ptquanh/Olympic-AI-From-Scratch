# 2. Dùng PyTorch nn.MultiheadAttention

```
import torch.nn as nn
# PyTorch MHA mặc định batch_first=False
mha = nn.MultiheadAttention(embed_dim=64, num_heads=4, batch_first=True)
x = torch.randn(2, 5, 64)
out, attn = mha(x, x, x)
print(out.shape)
```
