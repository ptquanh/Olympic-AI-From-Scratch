# Document AI Lab\nGiới thiệu pipeline OCR cơ bản. (Lưu ý: Bạn cần cài đặt pytesseract hoặc easyocr để chạy thực tế)

```
# Đây là mô phỏng pipeline xử lý Document AI
import numpy as np

def mock_ocr_pipeline(image_path):
    print(f"Loading image from {image_path}...")
    # Mô phỏng Bounding Boxes (x, y, w, h)
    boxes = [
        [10, 10, 100, 20],
        [10, 40, 200, 20]
    ]
    texts = [
        "TOTAL",
        "100.00 USD"
    ]
    return list(zip(boxes, texts))

results = mock_ocr_pipeline("invoice.jpg")
for box, text in results:
    print(f"Box: {box} -> Text: {text}")
```
