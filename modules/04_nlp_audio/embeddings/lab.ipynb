# Embeddings Lab\nThực hành tính Cosine Similarity giữa các vectors.

```
import torch
import torch.nn as nn
import torch.nn.functional as F

# Khởi tạo embedding layer (vocab=10, dim=4)
embedding = nn.Embedding(num_embeddings=10, embedding_dim=4)

# Lấy vector của từ index 2 và index 5
word1 = embedding(torch.tensor(2))
word2 = embedding(torch.tensor(5))

print("Word 1 vector:", word1.detach().numpy())
print("Word 2 vector:", word2.detach().numpy())

# Tính Cosine Similarity
cos_sim = F.cosine_similarity(word1.unsqueeze(0), word2.unsqueeze(0))
print("Cosine Similarity:", cos_sim.item())
```
