# Text Preprocessing Lab\nThực hành xử lý văn bản cơ bản bằng Regex và Tokenization.

```
import re

text = "Liên hệ với tôi qua email: example123@gmail.com hoặc truy cập https://google.com để biết thêm chi tiết!!! 😊"

# 1. Chuyển chữ thường
text = text.lower()
print("1. Lowercase:", text)

# 2. Xóa URL bằng Regex
text = re.sub(r"http\S+", "", text)
print("2. Xóa URL:", text)

# 3. Xóa Email
text = re.sub(r"\S+@\S+", "", text)
print("3. Xóa Email:", text)

# 4. Xóa Punctuation (Dấu câu)
text = re.sub(r"[^\w\s]", "", text)
print("4. Xóa Dấu câu:", text)

# 5. Xóa khoảng trắng thừa
text = re.sub(r"\s+", " ", text).strip()
print("5. Dọn khoảng trắng:", text)
```
