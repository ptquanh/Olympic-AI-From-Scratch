# Audio Fundamentals Lab\nThực hành xử lý âm thanh bằng torchaudio (vẽ waveform).

```
import torch
import numpy as np
import matplotlib.pyplot as plt

# Tạo một sóng sin nhân tạo (1 giây, 16kHz) thay vì load file để đảm bảo chạy được mà không cần dữ liệu
sample_rate = 16000
t = torch.linspace(0, 1, sample_rate)
freq = 440.0 # Tần số nốt A4
waveform = torch.sin(2 * np.pi * freq * t).unsqueeze(0)

print("Waveform shape:", waveform.shape)

# Vẽ đồ thị waveform (chỉ vẽ 0.01s đầu tiên để nhìn rõ sóng)
plt.figure(figsize=(10, 3))
plt.plot(t[:160].numpy(), waveform[0, :160].numpy())
plt.title("Waveform (440Hz Sine Wave - First 0.01s)")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```
