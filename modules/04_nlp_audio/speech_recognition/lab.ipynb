# Speech Recognition Metrics

WER/CER dùng edit distance; denominator là số token của reference.

```
def edit_distance(a, b):
    prev = list(range(len(b) + 1))
    for i, x in enumerate(a, 1):
        cur = [i]
        for j, y in enumerate(b, 1):
            cur.append(min(cur[-1] + 1, prev[j] + 1, prev[j - 1] + (x != y)))
        prev = cur
    return prev[-1]


def error_rate(reference, hypothesis, unit="word"):
    ref = reference.split() if unit == "word" else list(reference)
    hyp = hypothesis.split() if unit == "word" else list(hypothesis)
    if not ref:
        raise ValueError("reference must contain at least one unit")
    return edit_distance(ref, hyp) / len(ref)


ref = "the cat sat on the mat"
hyp = "the bat sat on the mat"
wer = error_rate(ref, hyp)
cer = error_rate(ref, hyp, "char")
assert abs(wer - 1 / 6) < 1e-12 and 0 <= cer <= 1
print("WER/CER", wer, cer)
```
