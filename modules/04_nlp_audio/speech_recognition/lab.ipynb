# Speech Recognition Lab\nThực hành tính Word Error Rate (WER).

```
# Cần cài đặt: pip install jiwer
try:
    import jiwer
    
    reference = "the cat sat on the mat"
    hypothesis = "the bat sat on the mat"
    
    # Tính WER
    wer = jiwer.wer(reference, hypothesis)
    print(f"Reference: {reference}")
    print(f"Hypothesis: {hypothesis}")
    print(f"Word Error Rate (WER): {wer * 100:.2f}%")
    
except ImportError:
    print("Vui lòng cài đặt thư viện jiwer bằng lệnh: pip install jiwer")
```
