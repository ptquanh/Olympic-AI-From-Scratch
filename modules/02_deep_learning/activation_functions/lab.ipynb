# Trực quan hóa các hàm kích hoạt
Sử dụng thư viện `matplotlib` để vẽ đồ thị.

```python
import matplotlib.pyplot as plt
import torch

x = torch.linspace(-5, 5, 100)
y_relu = torch.relu(x)

plt.plot(x.numpy(), y_relu.numpy())
plt.title('ReLU Activation')
plt.grid(True)
plt.show()
```
