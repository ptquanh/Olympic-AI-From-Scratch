# Hoạt động của Dropout ở chế độ Train và Eval

```python
import torch.nn as nn

dropout = nn.Dropout(p=0.5)
x = torch.ones(1, 10)

# Chế độ Train (Mặc định)
dropout.train()
print('Train Mode:', dropout(x))

# Chế độ Eval
dropout.eval()
print('Eval Mode:', dropout(x))
```
