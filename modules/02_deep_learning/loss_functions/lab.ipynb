# So sánh L1 Loss và L2 Loss (MSE)

```
import torch.nn as nn

pred = torch.tensor([1.0, 2.0, 3.0])
true = torch.tensor([1.0, 2.0, 10.0]) # Cố tình tạo 1 outlier

l1 = nn.L1Loss()
l2 = nn.MSELoss()
print('L1 Loss (MAE):', l1(pred, true).item())
print('L2 Loss (MSE):', l2(pred, true).item())
```
