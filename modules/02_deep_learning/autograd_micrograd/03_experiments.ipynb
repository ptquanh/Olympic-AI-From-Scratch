# Khảo sát: Cập nhật Gradient bị cộng dồn


```
w = torch.tensor(1.0, requires_grad=True)

for i in range(3):
    # Giả lập tính loss
    loss = w * 2
    loss.backward()
    print(f'Vòng {i+1}, w.grad = {w.grad.item()}')
    # CHÚ Ý: PyTorch cộng dồn gradient. Nếu không có lệnh w.grad.zero_() thì grad sẽ tăng mãi.
```
