# Experiment — accumulation

**Hypothesis:** gọi backward hai lần không zero gradient sẽ nhân đôi gradient.

```python
x=torch.tensor(2., requires_grad=True)
(x*x).backward(); once=x.grad.item(); (x*x).backward(); twice=x.grad.item()
assert twice == 2*once
x.grad.zero_(); (x*x).backward(); reset=x.grad.item()
print({"once":once,"twice":twice,"after_zero":reset})
```

**Observation:** gradient được cộng vào `.grad`; training loop phải reset trước backward mới.
