# 1. Tensor Basics
Tensor là cấu trúc dữ liệu cơ bản của PyTorch, tương tự như numpy.ndarray nhưng hỗ trợ tăng tốc GPU.

```
# Khởi tạo tensor
x = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)
print('Tensor:\n', x)
print('Shape:', x.shape)
```

# 2. Broadcasting & Operations
Thao tác nhân ma trận và thay đổi hình dạng (Reshape).

```
# Thay đổi hình dạng (Reshape / View)
x_flat = x.view(-1)
print('Flattened:', x_flat)
```
