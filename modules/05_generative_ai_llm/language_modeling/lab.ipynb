# Language Modeling & Text Generation

Trong lab này, chúng ta sẽ xem cách Tokenizer hoạt động và cách một mô hình sinh text (Causal Language Model) dự đoán từ tiếp theo.


```
# 1. Tokenizer
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("gpt2")
text = "Olympic AI is amazing!"
tokens = tokenizer.tokenize(text)
token_ids = tokenizer.encode(text)

print(f"Text: {text}")
print(f"Tokens: {tokens}")
```

## Text Generation Pipeline


```
# 2. Text Generation Pipeline
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")
prompt = "The future of AI in Vietnam is"
output = generator(prompt, max_length=30, num_return_sequences=1, truncation=True)

print("Generated text:")
```
