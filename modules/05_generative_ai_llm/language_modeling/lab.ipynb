# Language Modeling Lab — character bigram

CPU/offline model cho thấy train counts, next-token probabilities và perplexity mà không tải pretrained model.

```python
text="banana bandana"; vocab=sorted(set(text)); idx={c:i for i,c in enumerate(vocab)}; counts=np.ones((len(vocab),len(vocab)))
for a,b in zip(text,text[1:]): counts[idx[a],idx[b]]+=1
probs=counts/counts.sum(1,keepdims=True); nll=-np.mean([math.log(probs[idx[a],idx[b]]) for a,b in zip(text,text[1:])]); ppl=math.exp(nll)
assert probs.shape==(len(vocab),len(vocab)) and np.allclose(probs.sum(1),1) and ppl>=1
print("vocab",vocab,"perplexity",round(ppl,3),"next after a",{c:round(probs[idx['a'],idx[c]],3) for c in vocab})
```

Perplexity chỉ so sánh hợp lệ khi tokenization và evaluation data giống nhau; không trực tiếp đo factuality hoặc usefulness.
