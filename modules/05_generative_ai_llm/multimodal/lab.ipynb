# Multimodal Lab — align image/text embeddings

Fast/offline path mô phỏng hai encoder và cosine similarity; full model là bước optional có cache/license riêng.

```
rng = np.random.default_rng(42)
image_features = rng.normal(size=(3, 8))
text_features = image_features + rng.normal(0, 0.05, size=(3, 8))


def normalize(x):
    return x / np.linalg.norm(x, axis=1, keepdims=True)


similarity = normalize(image_features) @ normalize(text_features).T
pred = similarity.argmax(1)
assert np.array_equal(pred, np.arange(3))
print("similarity\n", np.round(similarity, 3), "matches", pred)
```

Cosine alignment là abstraction; VLM thật còn preprocessing ảnh, tokenizer, connector/projector và decoding. Không tải model ngầm trong smoke path.
