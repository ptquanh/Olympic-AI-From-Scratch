# Multimodal (VQA)

Sử dụng HuggingFace Pipeline cho bài toán Image Captioning và VQA.


```
import requests
from PIL import Image
from transformers import pipeline

# Load một ảnh từ internet
url = "https://huggingface.co/datasets/Narsil/image_dummy/resolve/main/parrots.png"
image = Image.open(requests.get(url, stream=True).raw)

# Hiển thị ảnh (nếu chạy trên Colab/Jupyter)
```

```
# 1. Image Captioning
captioner = pipeline("image-to-text", model="Salesforce/blip-image-captioning-base")
caption = captioner(image)
```

```
# 2. Visual Question Answering
vqa = pipeline("visual-question-answering", model="dandelin/vilt-b32-finetuned-vqa")
question = "What are the birds doing?"
answer = vqa(image=image, question=question)

print(f"Question: {question}")
```
