# Prompt Engineering

Thực hành Zero-shot, Few-shot và Chain-of-Thought.


```
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2") # Model quá nhỏ để làm CoT tốt, chỉ demo format

# 1. Zero-shot
zero_shot = "Translate English to French: hello =>"

# 2. Few-shot
few_shot = '''Translate English to French:
apple => pomme
cat => chat
hello =>'''

print("--- Zero Shot ---")
print(generator(zero_shot, max_new_tokens=5, pad_token_id=50256)[0]['generated_text'])

print("\n--- Few Shot ---")
```
