# Prompt Engineering Lab — prompt contract

Không cần gọi LLM. Ta xây prompt có task, input contract, output schema và test cases để đánh giá được.

```
def build_prompt(task,input_text,labels):
    return f"Task: {task}\nAllowed labels: {labels}\nInput: {input_text}\nOutput exactly one label; no explanation."
prompt=build_prompt("classify sentiment","dịch vụ rất tốt",["positive","negative"])
required=("Task:","Allowed labels:","Input:","Output")
assert all(item in prompt for item in required); print(prompt,"tokens_approx",len(prompt.split()))
```

```
test_cases=[("rất tốt","positive"),("quá tệ","negative"),("không tệ","positive")]
print("Evaluation contract:",{"cases":len(test_cases),"metric":"exact-match accuracy","edge_case":"negation"})
```
