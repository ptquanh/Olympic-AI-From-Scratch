# Experiment — rank và approximation error

**Hypothesis:** rank cao biểu diễn update tốt hơn nhưng tăng tham số.

```python
rng=np.random.default_rng(42); delta=rng.normal(size=(32,32)); u,s,vt=np.linalg.svd(delta,full_matrices=False)
for rank in (1,2,4,8,16):
    approx=(u[:,:rank]*s[:rank])@vt[:rank]; error=np.linalg.norm(delta-approx)/np.linalg.norm(delta)
    print(rank,"params",rank*64,"relative_error",round(error,4))
```

**Observation:** rank là trade-off capacity–memory; validation quyết định rank, không phải quy tắc cố định.
