Matrix shape, numerical derivative và descriptive statistics.

```python
A=np.array([[1.,2.],[3.,4.],[5.,6.]]); B=np.array([[7.,8.,9.],[10.,11.,12.]])
C=A@B; assert C.shape==(3,3); print(C)
```

```python
f=lambda x:x*x+2*x+1; x=2.; h=1e-5; numeric=(f(x+h)-f(x-h))/(2*h)
assert abs(numeric-6)<1e-8
scores=np.array([[8,7,9],[5,6,5],[9,9,10],[7,8,7],[6,5,6]])
print("gradient",numeric,"subject means",scores.mean(0),"student means",scores.mean(1))
```
