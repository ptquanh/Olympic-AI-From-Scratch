# Regex & Data Handling Lab

Pattern rõ mục tiêu; file demo dùng temporary directory nên không làm bẩn repo.

```python
import tempfile
text="Liên hệ a.nguyen@example.com (1999) hoặc team-ai@ptit.edu.vn (2005)."
emails=re.findall(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}",text)
years=re.findall(r"\b(?:19|20)\d{2}\b",text)
assert emails==["a.nguyen@example.com","team-ai@ptit.edu.vn"] and years==["1999","2005"]
print(emails,years)
```

```python
import tempfile
with tempfile.TemporaryDirectory() as tmp:
    root=Path(tmp); (root/"train").mkdir(); (root/"val").mkdir()
    for name in ("train/a.jpg","train/b.jpg","val/c.jpg"): (root/name).touch()
    files=sorted(p.relative_to(root).as_posix() for p in root.rglob("*.jpg"))
    assert files==["train/a.jpg","train/b.jpg","val/c.jpg"]
    print(files)
```
