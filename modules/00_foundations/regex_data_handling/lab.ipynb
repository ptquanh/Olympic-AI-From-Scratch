## Phần 1: Khởi động với Regex

Thử extract thông tin từ chuỗi lộn xộn dưới đây.

```python
text_data = """
Khách hàng 1: nguyen.van.a@example.com - SĐT: 0912345678 - Sinh năm: 1990
Khách hàng 2: tran_b_99@yahoo.com.vn - SĐT: 0987654321 - Sinh năm: 1999
Khách hàng 3: c.le@gmail.com - SĐT: 0123-456-789 - Sinh năm: 2005
"""

# TODO 1: Extract tất cả email
emails = re.findall(r'...', text_data) # Điền regex vào đây
print("Emails:", emails)

# TODO 2: Extract tất cả năm sinh (chỉ lấy 4 chữ số)
years = re.findall(r'...', text_data)
print("Years:", years)
```
