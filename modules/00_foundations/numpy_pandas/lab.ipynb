## Phần 1: Sức mạnh của Vectorization
Tạo 2 vector khổng lồ và so sánh tốc độ tính tổng phần tử tương ứng.

```python
N = 5_000_000
A = np.random.rand(N)
B = np.random.rand(N)

# 1. Vòng lặp For
start = time.time()
C_loop = np.zeros(N)
for i in range(N):
    C_loop[i] = A[i] + B[i]
print(f"For loop: {time.time() - start:.4f} giây")

# 2. Vectorization với NumPy
start = time.time()
C_numpy = A + B
print(f"NumPy vectorization: {time.time() - start:.4f} giây")

assert np.allclose(C_loop, C_numpy), "Kết quả không khớp!"
```
