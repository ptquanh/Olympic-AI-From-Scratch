```python
df=pd.DataFrame({"team":["a","a","b","b"],"score":[1.,np.nan,3.,5.]})
train=df.iloc[:3].copy(); valid=df.iloc[3:].copy(); fill=train["score"].median()
train["score"]=train["score"].fillna(fill); valid["score"]=valid["score"].fillna(fill)
summary=train.groupby("team",as_index=False)["score"].mean(); assert not train.isna().any().any()
print(summary,"validation",valid,sep="\n")
```
