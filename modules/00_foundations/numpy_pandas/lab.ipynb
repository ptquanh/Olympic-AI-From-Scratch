# NumPy & Pandas Lab

Vectorization, broadcasting và split-safe tabular preprocessing.

```
import pandas as pd

rng = np.random.default_rng(42)
X = rng.normal(size=(12, 3))
scale = np.array([1.0, 10.0, 100.0])
scaled = X * scale
assert scaled.shape == (12, 3)
print("column means", scaled.mean(0))
```

```
df = pd.DataFrame({"team": ["a", "a", "b", "b"], "score": [1.0, np.nan, 3.0, 5.0]})
train = df.iloc[:3].copy()
valid = df.iloc[3:].copy()
fill = train["score"].median()
train["score"] = train["score"].fillna(fill)
valid["score"] = valid["score"].fillna(fill)
summary = train.groupby("team", as_index=False)["score"].mean()
assert not train.isna().any().any()
print(summary, "validation", valid, sep="\n")
```
