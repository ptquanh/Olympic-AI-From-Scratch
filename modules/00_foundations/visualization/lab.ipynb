# Visualization Lab

Mỗi plot phải trả lời một câu hỏi và có title/axis/legend.

```
import pandas as pd

x = np.linspace(0, 2 * np.pi, 100)
fig, ax = plt.subplots()
ax.plot(x, np.sin(x), label="sin(x)")
ax.plot(x, np.cos(x), label="cos(x)")
ax.set(title="Hai tín hiệu tuần hoàn", xlabel="x (radian)", ylabel="amplitude")
ax.legend()
assert ax.get_title() and ax.get_xlabel() and ax.get_ylabel()
plt.close(fig)
print("plot contract passed")
```
