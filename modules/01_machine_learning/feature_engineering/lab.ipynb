# 🛠 Feature Engineering (Đặc trưng Dữ liệu)


Lab này sẽ tập trung vào tạo tính năng ngày tháng và mã hóa biến Categorical.

```
import pandas as pd
import numpy as np

data = {"Date": ["2023-01-01", "2023-01-02", "2023-01-03"], "City": ["HN", "HCM", "HN"], "Target": [1, 0, 1]}
df = pd.DataFrame(data)
```

```
# 1. Date Features
df["Date"] = pd.to_datetime(df["Date"])
df["Month"] = df["Date"].dt.month
df["DayOfWeek"] = df["Date"].dt.dayofweek
display(df)
```

```
# 2. One-Hot Encoding
df_encoded = pd.get_dummies(df, columns=["City"], drop_first=True)
display(df_encoded)
```
