# 2. Linear Regression (Framework)

Trong thực tế, bạn KHÔNG NÊN tự code lại Linear Regression (trừ khi để học). Thay vào đó, hãy dùng thư viện `scikit-learn` đã được tối ưu hóa cực kỳ tốt.

```
import numpy as np
import matplotlib.pyplot as plt
import time
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Tạo lại dữ liệu như bài From Scratch
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)
```

```
# Khởi tạo và huấn luyện mô hình bằng Sklearn
start_time = time.time()

model = LinearRegression()
model.fit(X, y)

end_time = time.time()
print(f'Training time: {(end_time - start_time)*1000:.2f} ms')

print(f'W (coef_): {model.coef_[0][0]:.4f}')
print(f'b (intercept_): {model.intercept_[0]:.4f}')

# Dự đoán và tính MSE
y_pred = model.predict(X)
mse = mean_squared_error(y, y_pred)
print(f'MSE: {mse:.4f}')
```

**Nhận xét:** Sklearn không dùng Gradient Descent vòng lặp for đơn giản như ta làm. Với Linear Regression, sklearn giải phương trình Normal Equation (tính toán đại số tuyến tính trực tiếp trên ma trận) nên ra kết quả chính xác tuyệt đối và tốc độ cực nhanh (thường < 1ms).
