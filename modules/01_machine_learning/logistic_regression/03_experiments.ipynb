# 3. Experiments: Phân loại Đa lớp (Multiclass)

Logistic Regression mặc định là Binary (0/1). Nếu có 3 class trở lên thì sao?
Sklearn hỗ trợ 2 chế độ:
- **One-vs-Rest (OvR):** Tạo 3 mô hình, mỗi mô hình dự đoán "Class A" vs "Không phải Class A".
- **Multinomial (Softmax):** Dùng hàm Softmax thay cho Sigmoid, dự đoán trực tiếp phân phối xác suất cho 3 class.

```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import make_blobs

# Tạo data 3 class
X, y = make_blobs(n_samples=300, centers=3, random_state=42, cluster_std=1.5)

models = {
    'One-vs-Rest': LogisticRegression(multi_class='ovr'),
    'Multinomial (Softmax)': LogisticRegression(multi_class='multinomial')
}

plt.figure(figsize=(10, 4))

for i, (name, clf) in enumerate(models.items()):
    clf.fit(X, y)
    
    plt.subplot(1, 2, i+1)
    
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.05),
                         np.arange(y_min, y_max, 0.05))
    
    Z = clf.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    
    plt.contourf(xx, yy, Z, alpha=0.3, cmap='viridis')
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis', edgecolor='k')
    plt.title(name)

plt.tight_layout()
plt.show()
```
