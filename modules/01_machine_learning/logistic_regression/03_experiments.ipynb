# Experiment — One-vs-Rest và multinomial

**Hypothesis:** hai strategy có decision boundary khác nhau trên bài toán ba lớp.

```python
from sklearn.datasets import make_blobs
from sklearn.linear_model import LogisticRegression
from sklearn.multiclass import OneVsRestClassifier
from sklearn.metrics import accuracy_score
X,y=make_blobs(n_samples=120 if FAST_MODE else 400,centers=3,random_state=42,cluster_std=1.5)
models={"ovr":OneVsRestClassifier(LogisticRegression(random_state=42,max_iter=300)),"multinomial":LogisticRegression(random_state=42,max_iter=300)}
for name,model in models.items(): model.fit(X,y); print(name,accuracy_score(y,model.predict(X)))
```

**Observation:** strategy là hyperparameter cần đánh giá trên validation; training accuracy ở đây chỉ là sanity check.
